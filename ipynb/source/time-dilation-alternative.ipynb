Using the cleaner relation:

$$
kc^2 = c^2 - v^2
$$

Divide both sides by \(c^2\):

$$
k = \frac{c^2 - v^2}{c^2}
$$

Split the fraction:

$$
k = \frac{c^2}{c^2} - \frac{v^2}{c^2}
$$

Therefore:

$$
k = 1 - \frac{v^2}{c^2}
$$

Now start from:

$$
t'^2 = \frac{s^2}{c^2 k}
$$

Since:

$$
c^2 = \frac{s^2}{t^2}
$$

we can rearrange this as:

$$
s^2 = c^2 t^2
$$

Substitute \(s^2 = c^2 t^2\) into:

$$
t'^2 = \frac{s^2}{c^2 k}
$$

This gives:

$$
t'^2 = \frac{c^2 t^2}{c^2 k}
$$

Cancel \(c^2\):

$$
t'^2 = \frac{t^2}{k}
$$

Now substitute:

$$
k = 1 - \frac{v^2}{c^2}
$$

So:

$$
t'^2 =
\frac{t^2}
{
1 - \frac{v^2}{c^2}
}
$$

Taking the square root gives:

$$
t' =
\frac{t}
{
\sqrt{1 - \frac{v^2}{c^2}}
}
$$

Therefore:

$$
t' = \gamma t
$$

where:

$$
\gamma =
\frac{1}
{
\sqrt{1 - \frac{v^2}{c^2}}
}
$$

So the final time-dilation relation is:

$$
t' =
\frac{t}
{
\sqrt{1 - \frac{v^2}{c^2}}
}
$$
