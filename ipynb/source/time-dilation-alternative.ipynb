This methodology implies that time dilation must also have a quadratic structure. Within this interpretation, light is represented as a kinetic object in an idealized system and is assigned an equivalent mass, denoted by $\tilde{m}$. The reduction in computational complexity can then be expressed through the corresponding equivalent kinetic-energy term of this idealized object.

---

The thought experiment considers a beam of light emitted from the rear of a train moving close to the speed of light. Because the speed of light does not combine with the speed of the train, the light does not gain the train’s additional velocity. From the idealized perspective of the model, the light therefore appears to lose part of its forward kinetic component relative to the train, causing it to fall behind the motion of the system. In computational relativity, the reduction in computational complexity can only be determined by examining the corresponding reduction in the object’s kinetic-energy term. In this case, the object is the idealized representation of light assigned an equivalent mass.

This apparent reduction can be used as a clock in itself. The time required for the light to traverse the moving system provides a direct measure of time dilation, without requiring a separate clock mechanism.

Therefore, there are two valid reasons for using the relation

$$
c^2-v^2.
$$

The first arises from the quadratic mathematical relationship between motion and the operation of the light clock that we retrieved from the K-e structure for the dilation of momentum through computational relativity. The second arises from the computational-relativity interpretation, in which energy represents combinatorial complexity and the term (v^2) describes the corresponding reduction in the system’s computational capacity.


Within computational relativity, we write

$$
k\left(\frac{1}{2}\tilde{m}c^2\right)
=

\frac{1}{2}\tilde{m}c^2
-
\frac{1}{2}\tilde{m}v^2,
$$

where (\tilde{m}) is an unknown constant mass-like parameter used to represent the energy of light. By convention, an energy-equivalent mass may be defined as

$$
m_{\mathrm{eq}}=\frac{E}{c^2},
$$

although (\tilde{m}) is not interpreted as the photon’s ordinary rest mass.

In this framework, energy is treated as combinatorial, so the subtracted term represents the portion of the system’s energy associated with computational or combinatorial complexity. The factor (k) therefore describes the remaining computational capacity:

$$
k
=
1-
\frac{v^2}{c^2}.
$$

Related ideas exist in effective-mass physics, information thermodynamics, and computational interpretations of physical systems, but this specific identification of relativistic dilation with the energy cost of combinatorial complexity is distinctive to computational relativity.
