In the **canonical Boltzmann distribution**, the Helmholtz free energy is directly related to the partition function \(Z\).

The probability of state \(i\) is

$$
\boxed{
p_i
=
\frac{e^{-E_i/(k_BT)}}{Z}
}
$$

where the partition function is

$$
\boxed{
Z
=
\sum_i e^{-E_i/(k_BT)}
}
$$

and

$$
\beta
=
\frac{1}{k_BT}.
$$

The **Helmholtz free energy** is defined as

$$
\boxed{
F
=
-k_BT\ln Z
}
$$

or equivalently,

$$
\boxed{
Z
=
e^{-F/(k_BT)}.
}
$$

This allows the Boltzmann probability to be written directly in terms of free energy:

$$
p_i
=
\frac{e^{-E_i/(k_BT)}}
{e^{-F/(k_BT)}}.
$$

Therefore,

$$
\boxed{
p_i
=
e^{-(E_i-F)/(k_BT)}
}
$$

The three quantities are therefore related by

$$
\boxed{
Z
=
\sum_i e^{-\beta E_i},
\qquad
F
=
-k_BT\ln Z,
\qquad
p_i
=
e^{-\beta(E_i-F)}.
}
$$

There is also the thermodynamic identity

$$
\boxed{
F
=
U-TS
}
$$

where \(U\) is the internal energy and \(S\) is the entropy.

Therefore, the partition function provides a bridge between the microscopic statistical description of the system and its macroscopic thermodynamic quantities:

$$
\boxed{
Z
\longrightarrow
F=-k_BT\ln Z
\longrightarrow
F=U-TS.
}
$$

If multiplicity or degeneracy \(\Omega_i\) is included, then the partition function becomes

$$
\boxed{
Z
=
\sum_i
\Omega_i
e^{-E_i/(k_BT)}
}
$$

and the probability of the corresponding macrostate is

$$
\boxed{
p_i
=
\frac{
\Omega_i e^{-E_i/(k_BT)}
}{
Z
}.
}
$$

This is where **multiplicity, the Boltzmann factor, the partition function \(Z\), and free energy all meet within the same statistical structure**.

---

### Equal-Energy Example

Suppose there are \(N\) accessible states, all having the same energy \(E\).

The partition function is then

$$
Z
=
Ne^{-E/(k_BT)}.
$$

The Helmholtz free energy is

$$
F
=
-k_BT\ln Z.
$$

Substituting the partition function,

$$
F
=
-k_BT
\ln\left(
Ne^{-E/(k_BT)}
\right).
$$

Using the logarithmic identity

$$
\ln(ab)
=
\ln a+\ln b,
$$

we obtain

$$
F
=
-k_BT
\left[
\ln N
-
\frac{E}{k_BT}
\right].
$$

Therefore,

$$
\boxed{
F
=
E-k_BT\ln N.
}
$$

If all \(N\) states are equally probable, the entropy is

$$
\boxed{
S
=
k_B\ln N.
}
$$

Therefore,

$$
TS
=
k_BT\ln N,
$$

and the free energy becomes

$$
\boxed{
F
=
E-TS.
}
$$

Thus, for this simple equal-energy system,

$$
\boxed{
\text{Free energy}
=
\text{energy}
-
\text{entropy contribution}.
}
$$

The role of \(N\) is important: increasing the number of accessible states increases the entropy,

$$
S=k_B\ln N,
$$

which correspondingly lowers the Helmholtz free energy through

$$
F
=
E-k_BT\ln N.
$$

Thus the multiplicity of the system enters the free-energy expression through the partition function \(Z\).

## Comparison of the Two Forms of Helmholtz Free Energy

The Helmholtz free energy can be expressed thermodynamically as

$$
\boxed{
F=U-TS
}
$$

and, for a canonical ensemble, statistically through the partition function as

$$
\boxed{
F=-k_BT\ln Z.
}
$$

These are not two different definitions of free energy. The statistical-mechanical expression reproduces the thermodynamic expression.

To demonstrate this, consider a simple system containing \(N=4\) accessible states, where every state has the same energy

$$
E=2k_BT.
$$

Because all four states have the same energy, the partition function is

$$
Z
=
\sum_{i=1}^{4}
e^{-E/(k_BT)}.
$$

Therefore,

$$
Z
=
4e^{-2}.
$$

Numerically,

$$
Z
\approx
4(0.135335)
=
0.54134.
$$

### Method 1: Free Energy from the Partition Function

Using

$$
F=-k_BT\ln Z,
$$

we obtain

$$
F
=
-k_BT\ln(0.54134).
$$

Since

$$
\ln(0.54134)
\approx
-0.61371,
$$

the free energy is

$$
\boxed{
F
\approx
0.61371\,k_BT.
}
$$

---

### Method 2: Free Energy from \(U-TS\)

Because every state has the same energy \(E=2k_BT\), the average internal energy is

$$
\boxed{
U=2k_BT.
}
$$

The four states are equally probable, so

$$
p_i=\frac14.
$$

The entropy is therefore

$$
S
=
-k_B
\sum_{i=1}^{4}
p_i\ln p_i.
$$

Substituting \(p_i=1/4\),

$$
S
=
-k_B
\left[
4
\left(
\frac14
\right)
\ln
\left(
\frac14
\right)
\right].
$$

This gives

$$
\boxed{
S=k_B\ln4
}
$$

or numerically,

$$
S
\approx
1.38629\,k_B.
$$

Therefore,

$$
TS
\approx
1.38629\,k_BT.
$$

Using the thermodynamic expression,

$$
F=U-TS,
$$

we obtain

$$
F
=
2k_BT
-
1.38629k_BT.
$$

Therefore,

$$
\boxed{
F
\approx
0.61371\,k_BT.
}
$$

---

### Comparison

Both approaches produce exactly the same result:

$$
\boxed{
-k_BT\ln Z
=
U-TS
=
0.61371\,k_BT.
}
$$

The equivalence can also be demonstrated algebraically. For \(N\) states all having the same energy \(E\),

$$
Z
=
Ne^{-E/(k_BT)}.
$$

Therefore,

$$
F
=
-k_BT
\ln
\left(
Ne^{-E/(k_BT)}
\right).
$$

Expanding the logarithm,

$$
F
=
-k_BT
\left[
\ln N
-
\frac{E}{k_BT}
\right],
$$

which gives

$$
\boxed{
F
=
E-k_BT\ln N.
}
$$

For this equal-energy system,

$$
U=E
$$

and

$$
S=k_B\ln N.
$$

Consequently,

$$
TS=k_BT\ln N,
$$

and therefore

$$
\boxed{
-k_BT\ln Z
=
E-k_BT\ln N
=
U-TS.
}
$$

Thus, the partition function \(Z\) contains the statistical information required to recover both the energy and entropy contributions to the Helmholtz free energy.
