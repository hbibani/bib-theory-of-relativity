
Start with the generalized binomial theorem:

$$
\boxed{
(1+x)^\alpha
=

\sum_{n=0}^{\infty}
\binom{\alpha}{n}x^n
}
$$

where

$$
\boxed{
\binom{\alpha}{n}
=

\frac{
\alpha(\alpha-1)(\alpha-2)\cdots(\alpha-n+1)
}{
n!
}
}
$$

Therefore,

$$
(1+x)^\alpha
=

1
+
\alpha x
+
\frac{\alpha(\alpha-1)}{2!}x^2
+
\frac{\alpha(\alpha-1)(\alpha-2)}{3!}x^3
+\cdots
$$

For the Lorentz factor,

$$
\boxed{
\gamma
=

\left(
1-\frac{v^2}{c^2}
\right)^{-1/2}
}
$$

we identify

$$
\boxed{
\alpha=-\frac12,
\qquad
x=-\frac{v^2}{c^2}.
}
$$

Hence

$$
\boxed{
\gamma
=

\sum_{n=0}^{\infty}
\binom{-\frac12}{n}
\left(
-\frac{v^2}{c^2}
\right)^n
}
$$

Expanding directly,

$$
\gamma
=

1
+
\left(-\frac12\right)
\left(-\frac{v^2}{c^2}\right)
+
\frac{
\left(-\frac12\right)
\left(-\frac32\right)
}{
2!
}
\left(-\frac{v^2}{c^2}\right)^2
+\cdots
$$

The signs cancel, giving

$$
\boxed{
\gamma
=

1
+
\frac12\frac{v^2}{c^2}
+
\frac38\frac{v^4}{c^4}
+
\frac5{16}\frac{v^6}{c^6}
+
\frac{35}{128}\frac{v^8}{c^8}
+
\frac{63}{256}\frac{v^{10}}{c^{10}}
+\cdots
}
$$

This can be written using the central binomial coefficient as

$$
\boxed{
\gamma
=

\sum_{n=0}^{\infty}
\frac{\binom{2n}{n}}{4^n}
\left(
\frac{v^2}{c^2}
\right)^n
}
$$

because

$$
\boxed{
\binom{-\frac12}{n}(-1)^n
=

\frac{\binom{2n}{n}}{4^n}.
}
$$

So equivalently,

$$
\boxed{
\gamma
=

\sum_{n=0}^{\infty}
\frac{\binom{2n}{n}}{4^n}
\frac{v^{2n}}{c^{2n}}
}
$$

---

For the inverse Lorentz factor,

$$
\boxed{
\frac{1}{\gamma}
=

\left(
1-\frac{v^2}{c^2}
\right)^{1/2}
}
$$

we now have

$$
\boxed{
\alpha=\frac12,
\qquad
x=-\frac{v^2}{c^2}.
}
$$

Therefore,

$$
\boxed{
\frac{1}{\gamma}
=

\sum_{n=0}^{\infty}
\binom{\frac12}{n}
\left(
-\frac{v^2}{c^2}
\right)^n
}
$$

Expanding,

$$
\frac{1}{\gamma}
=

1
+
\left(\frac12\right)
\left(-\frac{v^2}{c^2}\right)
+
\frac{
\left(\frac12\right)
\left(-\frac12\right)
}{
2!
}
\left(-\frac{v^2}{c^2}\right)^2
+\cdots
$$

which gives

$$
\boxed{
\frac{1}{\gamma}
=

1
-
\frac12\frac{v^2}{c^2}
-
\frac18\frac{v^4}{c^4}
-
\frac1{16}\frac{v^6}{c^6}
-
\frac5{128}\frac{v^8}{c^8}
-
\frac7{256}\frac{v^{10}}{c^{10}}
-\cdots
}
$$

The maximum value of $k_e$ is not fixed independently; it depends on which value of $n$ is chosen in




$$
\frac{mv^2}{2k_e}
=
\frac{\binom{2n}{n}}{4^n}.
$$

Solving for $k_e$,

$$
k_e
=
\frac{mv^2}
{2\left(\frac{\binom{2n}{n}}{4^n}\right)}.
$$

Therefore,

$$
\boxed{
k_e
=
\frac{4^n}{2\binom{2n}{n}}mv^2
}
$$

At the maximum velocity,

$$
v=c,
$$

so

$$
\boxed{
k_{e,\max}(n)
=
\frac{4^n}{2\binom{2n}{n}}mc^2
}
$$

Define

$$
a_n
=
\frac{\binom{2n}{n}}{4^n}.
$$

Then

$$
\boxed{
k_{e,\max}
=
\frac{1}{2a_n}mc^2
}
$$
