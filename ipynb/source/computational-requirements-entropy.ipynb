We retrieved this section from the experiment, that is, this formula was created after the investigation of the DA-anomoly prefix $\gamma$:

$$
p_{\text{obs}} = \gamma\, p_{\text{exp}}.
$$

After the experimental formula above, Boltzmann related probability to energy, and it turns out that this relation works very well with the formula above. Using the following formula, we can also relate probability dilation to energy, specifically kinetic energy.

Energy to Probability:

$$
p_{\text{exp}} \propto e^{-E_{\text{exp}}/(k_B T)}.
$$

Then Derive Dilation To Energy:

$$
\Delta E = -k_B T \ln\gamma
$$

From this relation, we can obtain a useful estimate of the expected temperature dilation by taking the derivative of the formula above. The derivative indicates that a certain amount of computational overhead is required to partition energy in a manner capable of producing a biological organism or another highly organized state.

Note: The bottom is the physical dilation factor (not the probability dilation $\gamma = \frac{p_{\text{obs}}}{ p_{\text{exp}}} $  )

$$\boxed{ \gamma_{d} \approx \frac{1}{1 - \frac{k_{b}T}{k_{e}\gamma}}}$$

 The condition on the value of $k_e$ follows from the requirement



  $$
  \boxed{
  k_e \gg \frac{k_B T}{\gamma}
  }
  $$

Beginning with the configurational energy associated with multiplicity,

$$
E_M
=

k_B T\ln M,
$$

taking the derivative with respect to multiplicity gives

$$
\frac{\partial E_M}{\partial M}
=

\frac{k_B T}{M}.
$$

This derivative represents the marginal energy required to increase the multiplicity of the configuration. It initially produces the dilation structure

$$
\frac{1}
{
1-\frac{k_B T}{M}
}.
$$

Because the term subtracted from (1) must be dimensionless, the derivative is divided by the reference energy constant (K_r):

$$
\frac{1}{K_r}
\frac{\partial E_M}{\partial M}
=

\frac{k_B T}{K_r M}.
$$

The resulting temperature-dilation relation is therefore

$$
\boxed{
T'
=

\frac{T}
{
1-\frac{k_B T}{K_r M}
}
}.
$$

This relation states that the marginal creation of multiplicity requires computational overhead relative to the available reference energy (K_r).
