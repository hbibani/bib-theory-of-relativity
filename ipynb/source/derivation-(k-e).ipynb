The following presents an alternative methodology for constructing a thought experiment that relates the internal motion of objects within a moving vehicle to the vehicle’s motion as measured by an external observer.


The thought experiment uses three distinct clocks.

## 1. External Observer's Clock

The stationary observer measures the train travelling a known distance (s) at velocity (v). The elapsed external time is

$$
t_{\mathrm{ext}}
=

\frac{s}{v}.
$$

This is the ordinary clock carried by the external observer.

## 2. Train-Motion Clock

The translational motion of the train is itself used as a clock. Each fixed distance travelled represents a fixed amount of elapsed time:

$$
t_{\mathrm{motion}}
=

\frac{s}{v}.
$$

Therefore,

$$
\boxed{
t_{\mathrm{motion}}
=
t_{\mathrm{ext}}
}
$$

The train-motion clock appears synchronized with the external clock because it is defined using the train's distance and velocity as measured by the external observer.

For example, if the train travels (100,\mathrm{m}) every second according to the external observer, then each (100,\mathrm{m}) interval functions as one external second.

## 3. Internal Clock of the Train

Start with

$$
v^2
=
v_{\mathrm{ext}}^2
-
\left(\frac{v^2}{c^2}\right)v_{\mathrm{ext}}^2.
$$

Factor (v_{\mathrm{ext}}^2):

$$
v^2
=
v_{\mathrm{ext}}^2
\left(1-\frac{v^2}{c^2}\right).
$$

Divide by the bracket:

$$
v_{\mathrm{ext}}^2
=
\frac{v^2}
{1-\frac{v^2}{c^2}}.
$$

Take the square root:

$$
\boxed{
v_{\mathrm{ext}}
=
\frac{v}
{\sqrt{1-\frac{v^2}{c^2}}}
}
$$
