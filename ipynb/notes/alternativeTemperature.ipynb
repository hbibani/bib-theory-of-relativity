### Corollary 3: The computational components for motion for objects with mass in the x, y, z direction

For motion, each perpendicular direction can be treated as having a computational component dedicated to motion along that axis. Consider a train moving close to the speed of light in the (x)-direction. The motion of particles within the train will then be affected differently depending on how their velocity is distributed across the (x), (y), and (z) directions.

If a particle moves primarily in the perpendicular (y)- or (z)-direction, its motion is affected differently from that of a particle whose velocity also contains a significant component in the horizontal (x)-direction. Within the computational interpretation, this occurs because part of the system's available computational capacity is already committed to maintaining the common horizontal motion of the train and everything contained within it.

As the horizontal velocity increases, a greater proportion of computational capacity is associated with the (x)-direction. This effectively draws computational complexity away from the perpendicular directions. Consequently, motion in the (y)- and (z)-directions is reduced in proportion to the horizontal velocity of the reference frame. The horizontal component therefore does not simply add to the perpendicular components; instead, it modifies the computational capacity available for perpendicular motion.

These computational components can be described through the effects of the associated dilation terms, which appear in both physical and thermodynamic systems within the framework of computational relativity.
